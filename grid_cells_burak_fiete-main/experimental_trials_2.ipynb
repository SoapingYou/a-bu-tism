```python
import numpy as np
from baseline_trials import run_sim_three
from RunSimulator import RunSimulator

baseline_tnum = 10
simlen = 20000

all_true_pos = np.zeros((baseline_tnum, simlen, 2))
all_can_sim_pos = np.zeros((baseline_tnum, simlen, 2))
all_base_sim_pos = np.zeros((baseline_tnum, simlen, 2))

run_sim_obj = RunSimulator()
#errors
simbase_final_drifts = np.zeros((baseline_tnum))
simbase_mean_errors = np.zeros((baseline_tnum,simlen))
simbase_rmses = np.zeros((baseline_tnum,simlen))
simbase_errors_over_time = [] #NOT a numpy array. its a 3d one, iter x simlen trial_num x floor(simlen/errorfreq)

simcan_final_drifts = np.zeros((baseline_tnum))
simcan_mean_errors = np.zeros((baseline_tnum))
simcan_rmses = np.zeros((baseline_tnum))
simcan_errors_over_time = [] #NOT a numpy array. its a 3d one, iter x simlen trial_num x floor(simlen/errorfreq)

for trial in range(baseline_tnum):
    result = run_sim_three(simlen)
    all_true_pos[trial] = result['true_pos']
    all_can_sim_pos[trial] = result['can_sim_pos'] 
    all_base_sim_pos[trial] = result['base_sim_pos']

    simbase_final_drifts[trial] = run_sim_obj.distance_error(all_true_pos[trial,-1,:],all_base_sim_pos[trial,-1,:] )
    simbase_mean_errors[trial] = run_sim_obj.mean_error(all_true_pos[trial,:,:],all_base_sim_pos[trial,:,:] )
    simbase_rmses[trial] = run_sim_obj.rmse(all_true_pos[trial,:,:],all_base_sim_pos[trial,:,:])
    simbase_errors_over_time.append(run_sim_obj.error_over_time(all_true_pos[trial,:,:], all_base_sim_pos[trial,:,:]))

    simcan_final_drifts[trial] = run_sim_obj.distance_error(all_true_pos[trial,-1,:],all_can_sim_pos[trial,-1,:] )
    simcan_mean_errors[trial] = run_sim_obj.mean_error(all_true_pos[trial,:,:],all_can_sim_pos[trial,:,:] )
    simcan_rmses[trial] = run_sim_obj.rmse(all_true_pos[trial,:,:],all_can_sim_pos[trial,:,:])
    simcan_errors_over_time.append(run_sim_obj.error_over_time(all_true_pos[trial,:,:], all_can_sim_pos[trial,:,:]))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x_labels = ["RMSE", "Final Drift"]
x_pos = np.arange(2)

avgcan_final_drifts = np.mean(simcan_final_drifts)
avgcan_mean_errors = np.mean(simcan_mean_errors.flatten())
avgcan_rmses = np.mean(simcan_rmses.flatten())

can_sim_std_final_drifts = np.std(simcan_final_drifts)
can_sim_std_mean_errors = np.std(simcan_mean_errors)
can_sim_std_rmses = np.std(simcan_rmses)

avgcanless_final_drifts = simbase_final_drifts
avgcanless_mean_errors = np.mean(simbase_mean_errors)
avgcanless_rmses = np.mean(simbase_rmses)

canless_sim_std_final_drifts = np.std(simbase_final_drifts)
canless_sim_std_mean_errors = np.std(simbase_mean_errors)
canless_sim_std_rmses = np.std(simbase_rmses)

width = 0.25

plt.figure(figsize=(12, 6))

plt.bar(x_pos[0]-width/2, avgcanless_final_drifts, width, label="Final Drift, baseline model", color="cornflowerblue", yerr=canless_sim_std_final_drifts, edgecolor="black")
plt.bar(x_pos[0]+width/2, avgcan_final_drifts, width, label="Final Drift, CAN model", color = "#0339C2", yerr=can_sim_std_final_drifts, edgecolor="black")
#plt.bar(x_pos, avg_mean_errors, width, label="Mean Error", color="mediumpurple", yerr=canless_sim_std_mean_errors, edgecolor="black")
plt.bar(x_pos[1] + width/2, avgcanless_rmses, width, label="RMSE, baseline model", color="#F852C0", yerr=canless_sim_std_rmses,edgecolor="black")
plt.bar(x_pos[1] + width, avgcan_rmses, width, label="RMSE, CAN model", color="#680848", yerr=can_sim_std_rmses,edgecolor="black")

plt.xticks(x_pos, x_labels, rotation=45)
plt.xlabel("Simulation Length")
plt.ylabel("Error Value")
plt.title("Average Drift and RMSE vs Simulation Length")

plt.legend()
plt.tight_layout()
plt.show()
```
